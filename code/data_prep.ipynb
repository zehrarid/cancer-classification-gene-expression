```python
print(golub.shape)
print(golub.head())
print(golub.columns[-10:].tolist())
print(golub.dtypes.value_counts())
```

```python
import requests
from pathlib import Path

url = "https://file.biolab.si/biolab/supp/bi-cancer/projections/_datasets/DLBCL.tab"

response = requests.get(url)
response.raise_for_status()

path = Path("../data/DLBCL.tab")
path.parent.mkdir(parents=True, exist_ok=True)
path.write_bytes(response.content)
```

```python
shipp = pd.read_csv(
    path,
    sep="\t",
    skiprows=[1, 2]
)

print(shipp.shape)
print(shipp.head())
```

```python
shipp.columns
```

```python
golub_probes = set(golub.columns[6:])
shipp_probes = set(shipp.columns[:-1])

common_probes = golub_probes & shipp_probes

print("Golub probes:", len(golub_probes))
print("Shipp probes:", len(shipp_probes))
print("Common probes:", len(common_probes))
```

```python
missing_from_shipp = golub_probes - shipp_probes
missing_from_golub = shipp_probes - golub_probes

print("Only in Golub:", missing_from_shipp)
print("Only in Shipp:", missing_from_golub)
```

```python
golub_only = sorted(golub_probes - shipp_probes)

control_probes = [p for p in golub_only if p.startswith("AFFX")]
other_probes = [p for p in golub_only if not p.startswith("AFFX")]

print("Control probes:", len(control_probes))
print("Other probes:", other_probes)
```

```python
shared = sorted(golub_probes & shipp_probes)

golub_expr = golub[shared].apply(pd.to_numeric)
shipp_expr = shipp[shared].apply(pd.to_numeric)

matches = []

for i in range(len(golub_expr)):
    for j in range(len(shipp_expr)):
        if np.array_equal(
            golub_expr.iloc[i].values,
            shipp_expr.iloc[j].values
        ):
            matches.append((i, j))

print("Exact matching profiles:", matches)
print("Number of exact matches:", len(matches))
```

```python
print(golub["Samples"].tolist())
print(shipp.index.tolist())
print(shipp["class"].value_counts())
```

```python
shared = sorted(set(golub_expr.columns) & set(shipp_expr.columns))

golub_common = golub_expr[shared]
shipp_common = shipp_expr[shared]

print("Golub range:")
print(golub_common.describe().loc[["min", "mean", "max"]].T.describe())

print("\nShipp range:")
print(shipp_common.describe().loc[["min", "mean", "max"]].T.describe())
```

```python
golub_means = golub_common.mean(axis=0)
shipp_means = shipp_common.mean(axis=0)

print(golub_means.corr(shipp_means))
```

```python
# Correlation between the average expression values of each shared probe
# (we already calculated this: 0.9475)

# Now compare the overall distributions
print("Golub overall:")
print(golub_common.stack().describe())

print("\nShipp overall:")
print(shipp_common.stack().describe())
```

```python
print(golub_means.corr(shipp_means, method="spearman"))
```
